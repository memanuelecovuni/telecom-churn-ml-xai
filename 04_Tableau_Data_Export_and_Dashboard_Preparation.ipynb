# 04 — Tableau Data Export and Dashboard Preparation

## Purpose

This notebook prepares validated analytical outputs from the preceding dissertation notebooks for visual communication in Tableau.

No models are trained, tuned or modified in this notebook. No new model-selection decisions are made. Instead, previously established dataset, predictive-performance and explainability results are reorganised into compact Tableau-ready datasets.

The planned Tableau visualisation is organised around three analytical themes:

1. **Dataset and Churn Profile** — communicates the characteristics of the telecommunications dataset, including class imbalance and relevant customer characteristics.
2. **Comparative Model Performance** — compares the five frozen imbalance-aware classifiers using held-out predictive metrics and confusion-matrix results.
3. **Explainability and Final Model Selection** — communicates cross-model feature importance, perturbation sensitivity and the evidence supporting the final selection of LightGBM.

The dashboards are intended as a visual communication layer over the validated dissertation analysis rather than an additional modelling or optimisation stage.

```python
# ============================================================
# Cell 5 — Reproduce Validated Notebook 1 Cleaning for Tableau
# ============================================================

import numpy as np
import pandas as pd


print("=== Tableau Customer Data Cleaning ===\n")


# ------------------------------------------------------------
# Work from a copy of the standardised raw dataset
# ------------------------------------------------------------

df_tableau_clean = (
    df_tableau_raw
    .copy()
)


# ------------------------------------------------------------
# 1. Correct known categorical typo
# ------------------------------------------------------------

sliver_before = int(
    (
        df_tableau_clean[
            "CRM_PID_Value_Segment"
        ] == "Sliver"
    ).sum()
)


df_tableau_clean[
    "CRM_PID_Value_Segment"
] = (
    df_tableau_clean[
        "CRM_PID_Value_Segment"
    ]
    .replace({
        "Sliver": "Silver"
    })
)


sliver_after = int(
    (
        df_tableau_clean[
            "CRM_PID_Value_Segment"
        ] == "Sliver"
    ).sum()
)


# ------------------------------------------------------------
# 2. Missing CRM segment -> Unknown
# ------------------------------------------------------------

crm_missing_before = int(
    df_tableau_clean[
        "CRM_PID_Value_Segment"
    ]
    .isna()
    .sum()
)


df_tableau_clean[
    "CRM_PID_Value_Segment"
] = (
    df_tableau_clean[
        "CRM_PID_Value_Segment"
    ]
    .fillna("Unknown")
)


# ------------------------------------------------------------
# 3. Reconstruct subscriber-count missingness
#
# Total_SUBs =
# Active + Not Active + Suspended
#
# Notebook 1 established this deterministic relationship.
# ------------------------------------------------------------

inactive_missing_before = int(
    df_tableau_clean[
        "Not_Active_subscribers"
    ]
    .isna()
    .sum()
)


suspended_missing_before = int(
    df_tableau_clean[
        "Suspended_subscribers"
    ]
    .isna()
    .sum()
)


# Missing inactive counts:
# infer using Total - Active - Suspended,
# treating missing suspended as zero for reconstruction.

inactive_mask = (
    df_tableau_clean[
        "Not_Active_subscribers"
    ]
    .isna()
)


df_tableau_clean.loc[
    inactive_mask,
    "Not_Active_subscribers"
] = (

    df_tableau_clean.loc[
        inactive_mask,
        "Total_SUBs"
    ]

    -

    df_tableau_clean.loc[
        inactive_mask,
        "Active_subscribers"
    ]

    -

    df_tableau_clean.loc[
        inactive_mask,
        "Suspended_subscribers"
    ]
    .fillna(0)
)


# Missing suspended counts can now be reconstructed
# from the same accounting identity.

suspended_mask = (
    df_tableau_clean[
        "Suspended_subscribers"
    ]
    .isna()
)


df_tableau_clean.loc[
    suspended_mask,
    "Suspended_subscribers"
] = (

    df_tableau_clean.loc[
        suspended_mask,
        "Total_SUBs"
    ]

    -

    df_tableau_clean.loc[
        suspended_mask,
        "Active_subscribers"
    ]

    -

    df_tableau_clean.loc[
        suspended_mask,
        "Not_Active_subscribers"
    ]
)


# ------------------------------------------------------------
# 4. Reconstruct the single missing ARPU value
#
# Notebook 1 established:
# TotalRevenue = 40.17
# Total_SUBs = 2
# ARPU = 20.085 -> 20.09
# ------------------------------------------------------------

arpu_missing_before = int(
    df_tableau_clean[
        "ARPU"
    ]
    .isna()
    .sum()
)


arpu_mask = (
    df_tableau_clean[
        "ARPU"
    ]
    .isna()
)


df_tableau_clean.loc[
    arpu_mask,
    "ARPU"
] = (

    df_tableau_clean.loc[
        arpu_mask,
        "TotalRevenue"
    ]

    /

    df_tableau_clean.loc[
        arpu_mask,
        "Total_SUBs"
    ]

).round(2)


# ------------------------------------------------------------
# 5. Recreate established descriptive ratio
# ------------------------------------------------------------

df_tableau_clean[
    "Inactive_Ratio"
] = np.where(

    df_tableau_clean[
        "Total_SUBs"
    ] > 0,

    df_tableau_clean[
        "Not_Active_subscribers"
    ]

    /

    df_tableau_clean[
        "Total_SUBs"
    ],

    0.0
)


# ------------------------------------------------------------
# Validation
# ------------------------------------------------------------

subscriber_identity_error = (

    df_tableau_clean[
        "Total_SUBs"
    ]

    -

    (
        df_tableau_clean[
            "Active_subscribers"
        ]

        +

        df_tableau_clean[
            "Not_Active_subscribers"
        ]

        +

        df_tableau_clean[
            "Suspended_subscribers"
        ]
    )
).abs()


print("=== Cleaning Audit ===")

print(
    "Sliver values corrected:",
    sliver_before
)

print(
    "Sliver values remaining:",
    sliver_after
)

print(
    "Missing CRM values filled:",
    crm_missing_before
)

print(
    "Missing inactive counts reconstructed:",
    inactive_missing_before
)

print(
    "Missing suspended counts reconstructed:",
    suspended_missing_before
)

print(
    "Missing ARPU values reconstructed:",
    arpu_missing_before
)


print(
    "\n=== Post-Cleaning Validation ==="
)

print(
    "Rows:",
    len(df_tableau_clean)
)

print(
    "Total missing cells:",
    int(
        df_tableau_clean
        .isna()
        .sum()
        .sum()
    )
)

print(
    "Maximum subscriber identity error:",
    float(
        subscriber_identity_error.max()
    )
)

print(
    "Minimum Inactive_Ratio:",
    float(
        df_tableau_clean[
            "Inactive_Ratio"
        ].min()
    )
)

print(
    "Maximum Inactive_Ratio:",
    float(
        df_tableau_clean[
            "Inactive_Ratio"
        ].max()
    )
)


# ------------------------------------------------------------
# Assertions
# ------------------------------------------------------------

assert len(
    df_tableau_clean
) == 8453

assert sliver_after == 0

assert (
    df_tableau_clean[
        "CRM_PID_Value_Segment"
    ]
    .isna()
    .sum()
    == 0
)

assert (
    df_tableau_clean[
        reconstruction_fields
    ]
    .isna()
    .sum()
    .sum()
    == 0
)

assert np.allclose(
    subscriber_identity_error,
    0
)

assert (
    df_tableau_clean[
        "ARPU"
    ]
    .isna()
    .sum()
    == 0
)

assert (
    df_tableau_clean[
        "CHURN"
    ]
    .value_counts()
    .get("Yes", 0)
    == 549
)


print(
    "\nTABLEAU_CUSTOMER_CLEANING_VALID: True"
)
```

```python
# ============================================================
# Cell 6 — Tableau Dashboard 1 Exports
# Dataset and Churn Profile
# ============================================================

import os
import pandas as pd


print("=== Dashboard 1: Dataset and Churn Profile ===\n")


# ------------------------------------------------------------
# 1. Dataset summary / KPI file
# ------------------------------------------------------------

total_customers = len(
    df_tableau_clean
)

churn_customers = int(
    (
        df_tableau_clean["CHURN"] == "Yes"
    ).sum()
)

non_churn_customers = int(
    (
        df_tableau_clean["CHURN"] == "No"
    ).sum()
)

churn_rate = (
    churn_customers
    / total_customers
)


d1_summary = pd.DataFrame([{

    "Total Customers":
        total_customers,

    "Churn Customers":
        churn_customers,

    "Non-Churn Customers":
        non_churn_customers,

    "Churn Rate":
        churn_rate,

    "Non-Churn Rate":
        non_churn_customers / total_customers,

    "Original Variables":
        14,

    "Model Predictors":
        9,

    "Final Encoded Inputs":
        24
}])


# ------------------------------------------------------------
# 2. Churn distribution
# ------------------------------------------------------------

d1_churn_distribution = (
    df_tableau_clean[
        "CHURN"
    ]
    .value_counts()
    .rename_axis(
        "Churn Status"
    )
    .reset_index(
        name="Customer Count"
    )
)


d1_churn_distribution[
    "Percentage"
] = (
    d1_churn_distribution[
        "Customer Count"
    ]
    / total_customers
)


# ------------------------------------------------------------
# 3. CRM value segment × churn
# ------------------------------------------------------------

d1_crm_segment = (
    df_tableau_clean
    .groupby(
        [
            "CRM_PID_Value_Segment",
            "CHURN"
        ],
        dropna=False
    )
    .size()
    .reset_index(
        name="Customer Count"
    )
    .rename(
        columns={
            "CRM_PID_Value_Segment":
                "CRM Value Segment",

            "CHURN":
                "Churn Status"
        }
    )
)


crm_totals = (
    d1_crm_segment
    .groupby(
        "CRM Value Segment"
    )[
        "Customer Count"
    ]
    .transform("sum")
)


d1_crm_segment[
    "Within-Segment Percentage"
] = (
    d1_crm_segment[
        "Customer Count"
    ]
    / crm_totals
)


# ------------------------------------------------------------
# 4. Effective segment × churn
# ------------------------------------------------------------

d1_effective_segment = (
    df_tableau_clean
    .groupby(
        [
            "EffectiveSegment",
            "CHURN"
        ],
        dropna=False
    )
    .size()
    .reset_index(
        name="Customer Count"
    )
    .rename(
        columns={
            "EffectiveSegment":
                "Effective Segment",

            "CHURN":
                "Churn Status"
        }
    )
)


effective_totals = (
    d1_effective_segment
    .groupby(
        "Effective Segment"
    )[
        "Customer Count"
    ]
    .transform("sum")
)


d1_effective_segment[
    "Within-Segment Percentage"
] = (
    d1_effective_segment[
        "Customer Count"
    ]
    / effective_totals
)


# ------------------------------------------------------------
# 5. Numerical profile by churn status
#
# Descriptive only — no new modelling/inference.
# ------------------------------------------------------------

profile_features = {

    "Active_subscribers":
        "Active Subscribers",

    "Not_Active_subscribers":
        "Inactive Subscribers",

    "Suspended_subscribers":
        "Suspended Subscribers",

    "AvgMobileRevenue":
        "Average Mobile Revenue",

    "AvgFIXRevenue":
        "Average Fixed Revenue",

    "ARPU":
        "ARPU",

    "Inactive_Ratio":
        "Inactive Subscriber Ratio"
}


profile_rows = []


for source_col, display_name in profile_features.items():

    for churn_status, group in (
        df_tableau_clean
        .groupby("CHURN")
    ):

        profile_rows.append({

            "Feature":
                display_name,

            "Churn Status":
                churn_status,

            "Customer Count":
                len(group),

            "Mean":
                group[source_col].mean(),

            "Median":
                group[source_col].median(),

            "Standard Deviation":
                group[source_col].std(),

            "Minimum":
                group[source_col].min(),

            "Maximum":
                group[source_col].max()
        })


d1_numeric_profile = pd.DataFrame(
    profile_rows
)


# ------------------------------------------------------------
# Export paths
# ------------------------------------------------------------

d1_files = {

    "tableau_d1_dataset_summary.csv":
        d1_summary,

    "tableau_d1_churn_distribution.csv":
        d1_churn_distribution,

    "tableau_d1_crm_segment_churn.csv":
        d1_crm_segment,

    "tableau_d1_effective_segment_churn.csv":
        d1_effective_segment,

    "tableau_d1_numeric_profile.csv":
        d1_numeric_profile
}


# ------------------------------------------------------------
# Write Tableau-ready CSVs
# ------------------------------------------------------------

for filename, df in d1_files.items():

    filepath = os.path.join(
        TABLEAU,
        filename
    )

    df.to_csv(
        filepath,
        index=False
    )


# ------------------------------------------------------------
# Validation
# ------------------------------------------------------------

assert int(
    d1_summary.loc[
        0,
        "Total Customers"
    ]
) == 8453


assert int(
    d1_summary.loc[
        0,
        "Churn Customers"
    ]
) == 549


assert (
    d1_churn_distribution[
        "Customer Count"
    ].sum()
    == 8453
)


assert (
    d1_crm_segment[
        "Customer Count"
    ].sum()
    == 8453
)


assert (
    d1_effective_segment[
        "Customer Count"
    ].sum()
    == 8453
)


assert len(
    d1_numeric_profile
) == 14


# ------------------------------------------------------------
# Report exports
# ------------------------------------------------------------

print("=== Dashboard 1 Export Audit ===\n")


audit_rows = []


for filename, df in d1_files.items():

    filepath = os.path.join(
        TABLEAU,
        filename
    )

    audit_rows.append({

        "File":
            filename,

        "Rows":
            len(df),

        "Columns":
            len(df.columns),

        "Missing Cells":
            int(
                df.isna()
                .sum()
                .sum()
            ),

        "Size (bytes)":
            os.path.getsize(
                filepath
            )
    })


d1_export_audit = pd.DataFrame(
    audit_rows
)


display(
    d1_export_audit
)


print(
    "\n=== Dataset Summary ===\n"
)

display(
    d1_summary
)


print(
    "\n=== Churn Distribution ===\n"
)

display(
    d1_churn_distribution
)


print(
    "\n=== CRM Segment × Churn ===\n"
)

display(
    d1_crm_segment
)


print(
    "\n=== Effective Segment × Churn ===\n"
)

display(
    d1_effective_segment
)


print(
    "\nTABLEAU_DASHBOARD_1_EXPORT_VALID: True"
)
```

```python
# ============================================================
# Cell 9 — Master Tableau Export Audit
# ============================================================

import os
import pandas as pd


print("=== MASTER TABLEAU EXPORT AUDIT ===\n")


# ------------------------------------------------------------
# Expected Tableau exports
# ------------------------------------------------------------

expected_exports = {

    # Dashboard 1
    "tableau_d1_dataset_summary.csv": (1, 8),
    "tableau_d1_churn_distribution.csv": (2, 3),
    "tableau_d1_crm_segment_churn.csv": (16, 4),
    "tableau_d1_effective_segment_churn.csv": (11, 4),
    "tableau_d1_numeric_profile.csv": (14, 8),

    # Dashboard 2
    "tableau_d2_model_performance.csv": (5, 13),
    "tableau_d2_metrics_long.csv": (35, 4),
    "tableau_d2_confusion_matrix.csv": (20, 5),

    # Dashboard 3
    "tableau_d3_explainability_ranks.csv": (96, 4),
    "tableau_d3_perturbation_sensitivity.csv": (25, 10),
    "tableau_d3_explainability_methods.csv": (5, 6),
    "tableau_d3_final_assessment.csv": (5, 14),
    "tableau_d3_selection_evidence.csv": (10, 3)
}


# ------------------------------------------------------------
# Load and audit every expected export
# ------------------------------------------------------------

loaded_exports = {}
audit_rows = []


for filename, expected_shape in expected_exports.items():

    filepath = os.path.join(
        TABLEAU,
        filename
    )

    exists = os.path.exists(
        filepath
    )

    assert exists, (
        f"Missing Tableau export: {filename}"
    )

    df = pd.read_csv(
        filepath
    )

    loaded_exports[
        filename
    ] = df

    actual_shape = df.shape

    shape_valid = (
        actual_shape
        == expected_shape
    )

    missing_cells = int(
        df.isna()
        .sum()
        .sum()
    )

    duplicate_rows = int(
        df.duplicated()
        .sum()
    )

    audit_rows.append({

        "Dashboard":
            filename.split("_")[2].upper(),

        "File":
            filename,

        "Rows":
            actual_shape[0],

        "Columns":
            actual_shape[1],

        "Expected Rows":
            expected_shape[0],

        "Expected Columns":
            expected_shape[1],

        "Shape Valid":
            shape_valid,

        "Missing Cells":
            missing_cells,

        "Duplicate Rows":
            duplicate_rows,

        "Size (bytes)":
            os.path.getsize(
                filepath
            )
    })


master_audit = pd.DataFrame(
    audit_rows
)


# ------------------------------------------------------------
# Dashboard 1 integrity checks
# ------------------------------------------------------------

d1_summary_check = loaded_exports[
    "tableau_d1_dataset_summary.csv"
]

d1_churn_check = loaded_exports[
    "tableau_d1_churn_distribution.csv"
]

d1_crm_check = loaded_exports[
    "tableau_d1_crm_segment_churn.csv"
]

d1_effective_check = loaded_exports[
    "tableau_d1_effective_segment_churn.csv"
]


assert int(
    d1_summary_check.loc[
        0,
        "Total Customers"
    ]
) == 8453


assert int(
    d1_summary_check.loc[
        0,
        "Churn Customers"
    ]
) == 549


assert int(
    d1_summary_check.loc[
        0,
        "Non-Churn Customers"
    ]
) == 7904


assert (
    d1_churn_check[
        "Customer Count"
    ].sum()
    == 8453
)


assert (
    d1_crm_check[
        "Customer Count"
    ].sum()
    == 8453
)


assert (
    d1_effective_check[
        "Customer Count"
    ].sum()
    == 8453
)


# ------------------------------------------------------------
# Dashboard 2 integrity checks
# ------------------------------------------------------------

d2_performance_check = loaded_exports[
    "tableau_d2_model_performance.csv"
]

d2_metrics_check = loaded_exports[
    "tableau_d2_metrics_long.csv"
]

d2_cm_check = loaded_exports[
    "tableau_d2_confusion_matrix.csv"
]


expected_models = {
    "Logistic Regression",
    "Decision Tree",
    "Random Forest",
    "XGBoost",
    "LightGBM"
}


assert set(
    d2_performance_check[
        "Model"
    ]
) == expected_models


assert set(
    d2_metrics_check[
        "Model"
    ]
) == expected_models


assert set(
    d2_cm_check[
        "Model"
    ]
) == expected_models


for model in expected_models:

    model_cm = d2_cm_check[
        d2_cm_check[
            "Model"
        ] == model
    ]

    assert (
        model_cm[
            "Count"
        ].sum()
        == 1691
    )

    assert (
        model_cm.loc[
            model_cm[
                "Actual Class"
            ] == "Churn",
            "Count"
        ].sum()
        == 110
    )

    assert (
        model_cm.loc[
            model_cm[
                "Actual Class"
            ] == "Non-Churn",
            "Count"
        ].sum()
        == 1581
    )


# ------------------------------------------------------------
# Dashboard 3 integrity checks
# ------------------------------------------------------------

d3_ranks_check = loaded_exports[
    "tableau_d3_explainability_ranks.csv"
]

d3_perturbation_check = loaded_exports[
    "tableau_d3_perturbation_sensitivity.csv"
]

d3_methods_check = loaded_exports[
    "tableau_d3_explainability_methods.csv"
]

d3_assessment_check = loaded_exports[
    "tableau_d3_final_assessment.csv"
]


# Four nonlinear models represented in SHAP rank comparison
expected_explained_models = {
    "Decision Tree",
    "Random Forest",
    "XGBoost",
    "LightGBM"
}


assert set(
    d3_ranks_check[
        "Model"
    ]
) == expected_explained_models


# All five models represented in perturbation analysis
assert set(
    d3_perturbation_check[
        "Model"
    ]
) == expected_models


assert set(
    d3_methods_check[
        "Model"
    ]
) == expected_models


assert set(
    d3_assessment_check[
        "Model"
    ]
) == expected_models


# Exactly one final selected model
selected_rows = (
    d3_assessment_check[
        d3_assessment_check[
            "Final Selected Model"
        ] == True
    ]
)


assert len(
    selected_rows
) == 1


assert (
    selected_rows[
        "Model"
    ].iloc[0]
    == "LightGBM"
)


# Average Mobile Revenue must be perturbation rank 1
# for all five models.

perturbation_rank_one = (
    d3_perturbation_check[
        d3_perturbation_check[
            "Sensitivity Rank"
        ] == 1
    ]
)


assert len(
    perturbation_rank_one
) == 5


assert set(
    perturbation_rank_one[
        "Feature"
    ]
) == {
    "Average Mobile Revenue"
}


# ------------------------------------------------------------
# Global export-quality checks
# ------------------------------------------------------------

assert master_audit[
    "Shape Valid"
].all()


assert (
    master_audit[
        "Missing Cells"
    ].sum()
    == 0
)


assert (
    master_audit[
        "Duplicate Rows"
    ].sum()
    == 0
)


# ------------------------------------------------------------
# Save audit itself
# ------------------------------------------------------------

audit_output_path = os.path.join(
    TABLEAU,
    "tableau_master_export_audit.csv"
)


master_audit.to_csv(
    audit_output_path,
    index=False
)


# ------------------------------------------------------------
# Display audit
# ------------------------------------------------------------

print(
    "=== Export File Audit ===\n"
)

display(
    master_audit
)


print(
    "\n=== Dashboard Summary ===\n"
)


dashboard_summary = (
    master_audit
    .groupby(
        "Dashboard"
    )
    .agg(
        Files=("File", "count"),
        Total_Rows=("Rows", "sum"),
        Missing_Cells=("Missing Cells", "sum"),
        Duplicate_Rows=("Duplicate Rows", "sum")
    )
    .reset_index()
)


display(
    dashboard_summary
)


# ------------------------------------------------------------
# Directory inventory
# ------------------------------------------------------------

print(
    "\n=== Final Tableau Directory Inventory ===\n"
)


tableau_inventory = []


for filename in sorted(
    os.listdir(TABLEAU)
):

    filepath = os.path.join(
        TABLEAU,
        filename
    )

    if os.path.isfile(
        filepath
    ):

        tableau_inventory.append({

            "File":
                filename,

            "Size (bytes)":
                os.path.getsize(
                    filepath
                )
        })


tableau_inventory = pd.DataFrame(
    tableau_inventory
)


display(
    tableau_inventory
)


# ------------------------------------------------------------
# Final report
# ------------------------------------------------------------

print(
    "\nExpected dashboard exports:",
    len(expected_exports)
)

print(
    "Validated dashboard exports:",
    len(master_audit)
)

print(
    "Total missing cells:",
    int(
        master_audit[
            "Missing Cells"
        ].sum()
    )
)

print(
    "Total duplicate rows:",
    int(
        master_audit[
            "Duplicate Rows"
        ].sum()
    )
)

print(
    "Final selected model:",
    selected_rows[
        "Model"
    ].iloc[0]
)

print(
    "Master audit saved:",
    audit_output_path
)


print(
    "\nTABLEAU_MASTER_EXPORT_AUDIT_VALID: True"
)
```
